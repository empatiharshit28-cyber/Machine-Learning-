```python
import pandas as pd

d = {
    'player': ['M S Dhoni', 'hardik', 'rahul', 'Mohit sharma', 'Andrerussel', 'jashprret', 'virat kohli', 'rohit sharma'],
    'Team': ['Chennai Super King', 'Mumbai indians', 'kings eleven', 'GT', 'kolkata knight riders', 'Mumbai indians', 'RCB', 'Mumbai indians'],
    'category': ['Batsman', 'Batsman', 'Batsman', 'Bowler', 'Bowler', 'Batsman', 'Batsman', 'Batsman'],
    'Bidprice': [13, 11, 10, 12, 7, 10, 17, 15],
    'Runs': [4000, 2400, 3000, 1000, 1500, 3500, 5000, 3900]
}

df = pd.DataFrame(d)
print(df)
```

```python
print(df[df['category'] == 'Bowler'])
```

```python
print(df[df['Team'] == 'Mumbai indians'])
```

```python
print(df[(df['Team'] == 'Mumbai indians') & (df['Runs'] > 3000)])
```

```python
print(df[df['Team'].isin(['RCB', 'Chennai Super King'])])
```

```python
print(df[['Bidprice', 'Runs']].mean())
```

```python
print(df.sort_values(by='Bidprice', ascending=False))
```

```python
print(df.loc[df['Bidprice'].idxmax()])
```

```python
print(df.loc[df['Runs'].idxmin()])
```

```python
print(len(df))
```

```python
print(df['Team'].value_counts())
```

```python
idx = df.groupby('Team')['Bidprice'].idxmax()
print(df.loc[idx])
```

```python
print(df.groupby('Team')['Runs'].mean())
```
